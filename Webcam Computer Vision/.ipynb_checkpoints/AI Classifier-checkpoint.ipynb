```python
import cv2
import os
import re
import pandas as pd
import cv2 # opencv library
import numpy as np
from os.path import isfile, join
import matplotlib.pyplot as plt
from datetime import datetime

```

```python
cars_cascade = cv2.CascadeClassifier('haarcascade_car.xml')
```

```python
col_frames = os.listdir('./images/')
analysis_img  = cv2.imread('./images/'+col_frames[2])
```

```python
plt.imshow(cv2.cvtColor(analysis_img, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
def detect_cars(frame):
    cars = cars_cascade.detectMultiScale(frame, 1.15, 3)
    print(str(cars.shape[0]), " vehicles detected")
    for (x, y, w, h) in cars:
        cv2.rectangle(frame, (x, y), (x+w,y+h), color=(0, 255, 0), thickness=2)
    return frame
```

```python
for x in range(-1,9):
    plt.imshow(cv2.cvtColor(detect_cars(cv2.imread('./images/'+col_frames[x])), cv2.COLOR_BGR2RGB))
    plt.show()

```

```python

plt.imshow(cv2.cvtColor(detect_cars(analysis_img), cv2.COLOR_BGR2RGB))
plt.show()
```
